This notebook demonstrates the complete data pipeline:
- Load SQLite database
- Execute SQL queries using pandas
- Reproduce JOIN using pandas merge
- Compare SQL JOIN and pandas merge results

```python
#import Libraries
import sqlite3
import pandas as pd
```

```python
#Connect Database
connection = sqlite3.connect("books.db")
```

```python
#First pd.read_sql()
query = """
SELECT title, rating
FROM books
WHERE rating >= 4;
"""

high_rating_books = pd.read_sql(query, connection)

high_rating_books.head()
```

```python
#Second pd.read_sql()
query = """
SELECT title, price_inr
FROM books
ORDER BY price_inr DESC
LIMIT 10;
"""

expensive_books = pd.read_sql(query, connection)

expensive_books
```

```python
#SQL JOIN
query = """
SELECT
books.title,
books.price_inr,
books.rating,
categories.category_name
FROM books
INNER JOIN categories
ON books.category_id = categories.category_id;
"""

sql_join = pd.read_sql(query, connection)

sql_join.head()
```

```python
#Pandas Merge
books_df = pd.read_sql(
    "SELECT * FROM books;",
    connection
)

categories_df = pd.read_sql(
    "SELECT * FROM categories;",
    connection
)

merged_df = pd.merge(
    books_df,
    categories_df,
    on="category_id",
    how="inner"
)

merged_df.head()
```

```python
#Compare Results
print("SQL JOIN Shape")

print(sql_join.shape)

print()

print("Pandas Merge Shape")

print(merged_df.shape)
```

```python
#Close Database
connection.close()
```
